# Sabse zyada bikne wala product


```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
print(product_sales)
```

# Sabse kam bikne wala product

```python
product_sales = product_sales.sort_values(ascending=True)
print(product_sales)
```

# Best performing city

```python
category_sales = df.groupby("City")["Sales"].sum()

category_sales = category_sales.sort_values(ascending=False)

print(category_sales)
```

# check Sales and Calculated Sales are same or not

```python
df["Calculated Sales"] = df["Quantity Ordered"] * df["Price Each"]

print(np.isclose(df["Sales"], df["Calculated Sales"]).all())
```

# Make new column of difference of sales and calculated sales 

```python
df["Difference"] = df["Sales"] - df["Calculated Sales"]
df[df["Difference"] != 0]
```

# Top 10 products of Bar Chart

```python
top_products = product_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top_products.index, top_products.values)
plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=75) 
plt.tight_layout()        
plt.show()

```

# City-wise sales ka Bar Chart

```python
plt.figure(figsize=(10,6))
plt.bar(category_sales.index, category_sales.values)
plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Month-wise Sales Trend (Line Chart)

```python

monthly_sales = df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(10,6))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```

# pie char of sales

```python
import matplotlib.pyplot as plt

pie_chart = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(8,8))

plt.pie(
    pie_chart,
    labels=pie_chart.index,
    autopct='%1.1f%%',
    startangle=140,
    labeldistance=1.1,
    pctdistance=0.74
)

plt.legend(
    pie_chart.index,
    title="Products",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.title("Sales Distribution by Product")

plt.tight_layout()
plt.show()
```
